# Object Oriented Programming

```python
class Car:
    def __init__(self, name, brand, model, vin):
        self.name = name
        self.brand = brand
        self.model = model
        self.vin = vin
    def describe(self):
        print(f"{self.name} is a {self.brand} {self.model}")
```

```python
car1 = Car("Taylor", "Honda", "Civic", "123456678")
```

```python
car2 = Car("Alex","Toyota","Corolla","132454512")
```

```python
car1.vin
```

```python
car2.vin
```

```python
car1.describe
```

```python
class Person:
    def __init__(self, first_name, middle_name, last_name, year_of_birth):
        self.first_name = first_name
        self.middle_name = middle_name
        self.last_name = last_name 
        self.year_of_birth = year_of_birth
    
    def full_name(self):
        print(f"{self.first_name} {self.last_name}")

    def initials(self):
        print(f"{self.first_name[0]}{self.middle_name[0]}{self.last_name[0]}")
```

```python
my_attributes = Person ("John", "Aaron", "Smith", 1947)
```

```python
my_attributes.full_name()
```

```python
my_attributes.initials()
```

# Privacy in Python

```python
class Car:
    def __init__(self, name, brand, model, vin):
        self.name = name
        self.brand = brand
        self.model = model
        self.__vin = vin # this will make python use a getter to get the vin
    def get_vin(self):
        return self.__vin
    
    def set_vin(self, new_vin):
        if len(self.new_vin != 17):
            print ("invalid VIN length")
        else: 
            self.__vin = new_vin 
```

```python
car1 = Car("Taylor", "Honda", "Civic", "1231234523543")
```

```python
car1.get_vin() # this is us running the getter here.
```

```python
car1.set_vin('12345678912345678')
```

1. Create a class called `Book` that has the attributes title, author, pages, price. It also has the method get_price() which returns price, and set_price() to ensure price is numeric.

Then create two instances of this class:

b1:
* title -> 'Practical Programming'
* author -> 'Gries, Campbell, Montojo'
* pages -> 383
* price -> 50

b2:
* title -> 'Building a Career in Data Science'
* author -> 'Robinson, Nolis'
* pages -> 322
* price -> 40

```python
class Book:
    def __init__(self, title, author, pages, price):
        self.title = title
        self.author = author
        self.pages = pages
        self.__price = price
    
    def get_price(self):
        return self.__price
    
    def set_price(self, new_price):
        if type(new_price) == float or type(new_price) == int:
            self.__price = new_price
        else: 
            print ("Please ensure you are entering a number for the price")


    def total_price(self):
        return (round(self.__price * 1.13))
```

```python
b1 = Book("Practical Programming", "Gries, Campbell, Montojo", 383, 50)
b2 = Book("Building a Career in Data Science", "Robinson, Nolis", 322, 40)
```
